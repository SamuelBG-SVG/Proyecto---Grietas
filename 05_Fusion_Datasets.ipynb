```python
import os
import shutil

# Directorio base local
origen_base = '/content/temp_sdnet/'

# Directorios de destino(Google Drive)
dest_positive = '/content/drive/MyDrive/Proyecto_Grietas/data/Positive/'
dest_negative = '/content/drive/MyDrive/Proyecto_Grietas/data/Negative/'

os.makedirs(dest_positive, exist_ok=True)
os.makedirs(dest_negative, exist_ok=True)

# Las carpetas internas que contienen las clases
pos_folders = ['CP', 'CW', 'CD']
neg_folders = ['UP', 'UW', 'UD']

added_pos = 0
added_neg = 0

print("🔍 Recorriendo P, D y W para unir las imágenes...")

for root, dirs, files in os.walk(origen_base):
    current_folder_name = os.path.basename(root)

    # Si la carpeta actual es positiva (CP, CW, CD)
    if current_folder_name in pos_folders:
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                src_file = os.path.join(root, file)
                dst_file = os.path.join(dest_positive, f"sdnet_{current_folder_name}_{file}")
                if not os.path.exists(dst_file):
                    shutil.copy(src_file, dst_file)
                    added_pos += 1

    # Si la carpeta actual es negativa (UP, UW, UD)
    elif current_folder_name in neg_folders:
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                src_file = os.path.join(root, file)
                dst_file = os.path.join(dest_negative, f"sdnet_{current_folder_name}_{file}")
                if not os.path.exists(dst_file):
                    shutil.copy(src_file, dst_file)
                    added_neg += 1

print(f"\n🎉 ¡Fusión local completada con éxito!")
print(f"• Nuevas imágenes añadidas a Positive: {added_pos}")
print(f"• Nuevas imágenes añadidas a Negative: {added_neg}")
print(f"• Total final en Positive: {len(os.listdir(dest_positive))}")
print(f"• Total final en Negative: {len(os.listdir(dest_negative))}")
```
