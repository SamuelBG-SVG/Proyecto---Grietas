```python
import os

ruta_data = "/content/drive/MyDrive/Proyecto_Grietas/data"

ruta_positive = os.path.join(ruta_data, "Positive")
ruta_negative = os.path.join(ruta_data, "Negative")

print("Positive:", len(os.listdir(ruta_positive)))
print("Negative:", len(os.listdir(ruta_negative)))
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import random
import os


positive_files = os.listdir(ruta_positive)
negative_files = os.listdir(ruta_negative)

plt.figure(figsize=(10, 8))

for i in range(4):
    archivo = random.choice(positive_files)
    imagen = Image.open(os.path.join(ruta_positive, archivo))

    plt.subplot(2, 4, i + 1)
    plt.imshow(imagen)
    plt.title("Positive")
    plt.axis("off")

for i in range(4):
    archivo = random.choice(negative_files)
    imagen = Image.open(os.path.join(ruta_negative, archivo))

    plt.subplot(2, 4, i + 5)
    plt.imshow(imagen)
    plt.title("Negative")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
imagen_prueba = Image.open(
    os.path.join(ruta_positive, positive_files[0])
)

print("Resolución:", imagen_prueba.size)
print("Formato:", imagen_prueba.format)
```
